### Preparing the dataset

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("car_fuel_efficiency_2026.csv")
df.head()
```

```python
# Fill missing values with zeros
df = df.fillna(0)
df.info()
```

```python
# Do the split with these exact calls
from sklearn.model_selection import train_test_split

df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=1
)

# Prepare target variable
y_train = df_train.fuel_efficiency_mpg.values
y_val = df_val.fuel_efficiency_mpg.values
del df_train['fuel_efficiency_mpg']
del df_val['fuel_efficiency_mpg']
```

```python
# Use DictVectorizer(sparse=True) to turn the dataframes into matrices
from sklearn.feature_extraction import DictVectorizer

# Convert each row into dict
train_dicts = df_train.to_dict(orient='records')
val_dicts = df_val.to_dict(orient='records')

dv = DictVectorizer(sparse=False)

X_train = dv.fit_transform(train_dicts)
X_val = dv.transform(val_dicts)
```

### Q1

```python
# Train a decision tree regressor to predict the fuel_efficiency_mpg variable.
# Train a model with max_depth=1

from sklearn.tree import DecisionTreeRegressor

dt = DecisionTreeRegressor(max_depth=1)
dt.fit(X_train, y_train)
```

```python
# Which original feature is used for splitting the data? If you use DictVectorizer, map a one-hot feature such as origin=USA back to its original column name.
feature_idx = dt.tree_.feature[0]
dv.get_feature_names_out()[feature_idx]
```

### Q2

```python
# Train a random forest regressor with these parameters
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=10, random_state=1, n_jobs=-1)
rf.fit(X_train, y_train)
```

```python
# What's the RMSE of this model on the validation data?
from sklearn.metrics import root_mean_squared_error

y_pred = rf.predict(X_val)
rmse = root_mean_squared_error(y_val, y_pred)
print("RMSE:", rmse)
```

### Q3

```python
estimators = [10, 50, 100, 150]
random_state= 1

for n in estimators:
    rf = RandomForestRegressor(n_estimators=n, random_state=random_state, n_jobs=-1)
    rf.fit(X_train, y_train)

    y_pred = rf.predict(X_val)
    rmse = root_mean_squared_error(y_val, y_pred)
    print(f"Estimators {n}. RMSE: {rmse}")
```

### Q4

```python
# Try different values of max_depth: [10, 15, 20, 25]
# For each:
#   try different values of n_estimators from [10, 50, 100, 150]
#   calculate the mean RMSE

max_depths = [10, 15, 20, 25]
estimators = [10, 50, 100, 150]
random_state= 1
scores = []

for m in max_depths:
    for n in estimators:
        rf = RandomForestRegressor(n_estimators=n, random_state=random_state, n_jobs=-1, max_depth=m)
        rf.fit(X_train, y_train)

        y_pred = rf.predict(X_val)
        rmse = root_mean_squared_error(y_val, y_pred)
        scores.append((m, n, rmse))

df_scores = pd.DataFrame(scores, columns=['max_depth', 'n_estimators', 'rmse'])
df_scores.pivot(
    index='max_depth',
    columns='n_estimators',
    values='rmse'
).round(3)
print(df_scores)
```

### Q5

```python
# Get the feature importance information from this model

rf = RandomForestRegressor(n_estimators=10, random_state=1, n_jobs=-1, max_depth=20)
rf.fit(X_train, y_train)

features = dv.get_feature_names_out()
importances = rf.feature_importances_

df_importance = pd.DataFrame({
    'feature': features,
    'importance': importances
})

df_importance.sort_values('importance', ascending=False).head(10)
```

### Q6

```python
import xgboost as xgb

features = list(dv.get_feature_names_out())
dtrain = xgb.DMatrix(X_train, label=y_train, feature_names=features)
dval = xgb.DMatrix(X_val, label=y_val, feature_names=features)

watchlist = [(dtrain, 'train'), (dval, 'val')]

xgb_params = {
    'eta': 0.3,  # Now change eta from 0.3 to 0.1
    'max_depth': 6,
    'min_child_weight': 1,
    
    'objective': 'reg:squarederror',
    'nthread': 8,
    
    'seed': 1,
    'verbosity': 1,
}

model = xgb.train(
    xgb_params,
    dtrain,
    num_boost_round=100,
    evals=watchlist,
    verbose_eval=10
)

y_pred = model.predict(dval)
rmse_03 = root_mean_squared_error(y_val, y_pred)
print('RMSE eta=0.3:', rmse_03)
```

```python
xgb_params['eta'] = 0.1

model = xgb.train(
    xgb_params,
    dtrain,
    num_boost_round=100,
    evals=watchlist,
    verbose_eval=10
)

y_pred = model.predict(dval)

rmse_01 = root_mean_squared_error(y_val, y_pred)

print('RMSE eta=0.1:', rmse_01)
```
